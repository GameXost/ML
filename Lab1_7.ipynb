3. Создайте матрицу 8х8 и заполните ее паттерном шахматной доски

````python
desk = np.zeros((8, 8))
for x in range(8):
  for y in range(8):
    desk[x][y] = 0 if (x + y) % 2 == 0 else 1
print(desk)
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python

df["Chronic Diseases"].value_counts(dropna=False)
````

````python
bmi = df['BMI']

q30 = bmi.quantile(0.30)
q60 = bmi.quantile(0.60)
q75 = bmi.quantile(0.75)

up = bmi > q75
mid = (bmi <= q60) & (bmi >= q30)

disease = df['Chronic Diseases'].notna()

print(disease[up].mean() * 100)
print(disease[mid].mean() * 100)

# чаще
````

7. Красиво визуализируйте как связаны между собой уровень дохода, когнитивные способности и уровень стресса*

\* Задание на доп. балл

````python
plt.figure(figsize=(10, 6))

stress = df['Stress Levels']
cognitive = df['Cognitive Function']
income = df['Income Level']

for lvl in ['Low', 'Medium', 'High']:
  plt.scatter(
      stress[income == lvl],
      cognitive[income == lvl],
      label = lvl,
      alpha = 0.4,
      s = 20
  )
plt.title('Cognitive Function, stress and income')
plt.xlabel('Stress Level')
plt.ylabel('Cognitive Function')
plt.legend(title = 'Income level')
plt.grid(alpha = 0.5)
plt.show()
````

````python
fig, axes = plt.subplots(1, 3, figsize = (15, 5), sharex = True, sharey = True)

for axe, lvl in zip(axes, ['Low', 'Medium', 'High']):
  axe.scatter(stress[income == lvl], cognitive[income==lvl], alpha = 0.3, s=20)
  axe.set_title(f'Income {lvl}')
  axe.set_xlabel('Stress Level')
  axe.grid(alpha = 0.5)

axes[0].set_ylabel("Cognitive Function")
plt.show()
````
